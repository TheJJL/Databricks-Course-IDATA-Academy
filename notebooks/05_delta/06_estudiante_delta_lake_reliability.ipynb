## 11. Comparar versión actual vs versión anterior

Esta consulta muestra la diferencia entre el estado inicial y el estado actual de la orden.

````python
%sql
WITH version_inicial AS (
  SELECT
    order_id,
    order_status AS order_status_version_0,
    delay_days AS delay_days_version_0
  FROM workspace.delta_lab.orders_reliability_lab VERSION AS OF 0
  WHERE order_id = (SELECT order_id FROM v_order_lab_candidate)
), version_actual AS (
  SELECT
    order_id,
    order_status AS order_status_actual,
    delay_days AS delay_days_actual
  FROM workspace.delta_lab.orders_reliability_lab
  WHERE order_id = (SELECT order_id FROM v_order_lab_candidate)
)
SELECT
  a.order_id,
  i.order_status_version_0,
  a.order_status_actual,
  i.delay_days_version_0,
  a.delay_days_actual
FROM version_actual a
JOIN version_inicial i
  ON a.order_id = i.order_id;
````

### Análisis Reto Nivel 1: Historia de tabla

**Tabla analizada:** `workspace.delta_lab.orders_reliability_lab`

**Hallazgos:**

* **Versión inicial:** 0 (operación: CREATE OR REPLACE TABLE)
* **Última versión:** 2 (operación: RESTORE)
* **Última operación:** RESTORE TABLE TO VERSION AS OF 0
* **Evidencia más importante:**
  - Version 0: Tabla creada con 99,441 filas desde orders_clean
  - Version 1: UPDATE modificó 1 fila (order_status → 'canceled')
  - Version 2: RESTORE recuperó estado original completo

**Conclusión:** Delta Lake registra **todas** las operaciones. Incluso un UPDATE de 1 fila queda auditado. RESTORE no borra el historial, crea una nueva versión que apunta a un estado anterior.

````python
%sql
-- RETO NIVEL 2: Time Travel y comparación entre versión 0 y versión 1
-- Vamos a comparar el estado antes y después del UPDATE en bagazo_reliability_lab

WITH version_0 AS (
  SELECT fecha, ingenio, lluvia_mm AS lluvia_v0, cana_molida_ton, bagazo_entregado_ton
  FROM workspace.delta_lab.bagazo_reliability_lab VERSION AS OF 0
  WHERE ingenio = 'Providencia'
  ORDER BY fecha
  LIMIT 1
),
version_1 AS (
  SELECT fecha, ingenio, lluvia_mm AS lluvia_v1, cana_molida_ton, bagazo_entregado_ton
  FROM workspace.delta_lab.bagazo_reliability_lab VERSION AS OF 1
  WHERE ingenio = 'Providencia'
  ORDER BY fecha
  LIMIT 1
),
version_actual AS (
  SELECT fecha, ingenio, lluvia_mm AS lluvia_actual, cana_molida_ton, bagazo_entregado_ton
  FROM workspace.delta_lab.bagazo_reliability_lab
  WHERE ingenio = 'Providencia'
  ORDER BY fecha
  LIMIT 1
)
SELECT
  v0.fecha,
  v0.ingenio,
  v0.lluvia_v0 AS lluvia_version_0_original,
  v1.lluvia_v1 AS lluvia_version_1_corrupta,
  va.lluvia_actual AS lluvia_version_actual_restaurada,
  CASE
    WHEN v0.lluvia_v0 = va.lluvia_actual THEN '✅ RESTAURADO CORRECTAMENTE'
    ELSE '❌ DIFERENTE'
  END AS estado_restauracion
FROM version_0 v0
INNER JOIN version_1 v1 ON v0.fecha = v1.fecha AND v0.ingenio = v1.ingenio
INNER JOIN version_actual va ON v0.fecha = va.fecha AND v0.ingenio = va.ingenio;
````

### Análisis Reto Nivel 2: Time Travel y comparación

**¿Qué cambió entre versiones?**

* **Versión 0 (original):** `lluvia_mm` tenía valor válido (e.g., 5.07 mm)
* **Versión 1 (corrupta):** `lluvia_mm` cambió a `-10` (valor inválido intencional)
* **Versión 2 (restaurada):** `lluvia_mm` volvió al valor original de la versión 0

**Lecciones clave:**

1. **VERSION AS OF permite auditar cualquier punto en el tiempo** — Podemos ver exactamente qué valor tenía la columna antes, durante y después de la corrupción
2. **RESTORE no borra historial** — La versión 1 (corrupta) sigue existiendo en el historial para auditoría
3. **Recuperación instantánea** — RESTORE recuperó el estado original sin pérdida de datos
4. **Detección de anomalías** — `WHERE lluvia_mm < 0` detectó el valor inválido inmediatamente

**Valor para el negocio:** En producción, si un error de ingesta escribe valores inválidos, podemos:
- Detectarlo con reglas de calidad (`lluvia_mm < 0`)
- Auditar cuándo ocurrió con `DESCRIBE HISTORY`
- Recuperar datos válidos con `RESTORE` en segundos, sin pérdida de información
